```python
# Taller de Computación Visual - Punto 11
# Implementación de Matrices de Proyección Ortográfica y Perspectiva con NumPy.

import numpy as np

# --- 1. Coordenadas Homogéneas ---
p_mundo = np.array([5, 2, -10, 1])

print("--- PUNTO 3D EN EL MUNDO ---")
print(f"Punto 3D (x, y, z): [5, 2, -10]")
print(f"Punto Homogéneo (x, y, z, w): {p_mundo}")


# --- 2. Matriz de Proyección Ortográfica ---
l, r = -10, 10  # left, right
b, t = -10, 10  # bottom, top
n, f = 1, 50    # near, far (distancias desde la cámara)

# Fórmula de la matriz Ortográfica
M_ortho = np.array([
    [2/(r-l), 0,       0,         -(r+l)/(r-l)],
    [0,       2/(t-b), 0,         -(t+b)/(t-b)],
    [0,       0,       -2/(f-n),  -(f+n)/(f-n)], # Mapea -Z
    [0,       0,       0,         1]
])

p_clip_ortho = M_ortho @ p_mundo
p_ndc_ortho = p_clip_ortho[:3] / p_clip_ortho[3]

print("\n--- PROYECCIÓN ORTOGRÁFICA ---")
print(f"Punto en 'Clip Space': {p_clip_ortho}")
print(f"Coordenadas de Pantalla (NDC): {p_ndc_ortho}")


# --- 3. Matriz de Proyección Perspectiva ---
fov = np.radians(60)  # Campo de visión (Field of View) de 60 grados
aspecto = 16 / 9      # Relación de aspecto (pantalla ancha)
n_persp = 0.1         # Plano cercano
f_persp = 100.0       # Plano lejano

# Cálculo para la matriz (t = 1/tan(fov/2))
t_calc = 1.0 / np.tan(fov / 2.0)

# Fórmula de la matriz de Perspectiva
M_persp = np.array([
    [t_calc/aspecto, 0,      0,                               0],
    [0,              t_calc, 0,                               0],
    [0,              0,      (f_persp+n_persp)/(n_persp-f_persp), (2*f_persp*n_persp)/(n_persp-f_persp)],
    [0,              0,      -1,                              0]
])

p_clip_persp = M_persp @ p_mundo

print("\n--- PROYECCIÓN PERSPECTIVA ---")
print(f"Punto en 'Clip Space' (antes de dividir): {p_clip_persp}")

w = p_clip_persp[3]
p_ndc_persp = p_clip_persp[:3] / w

print(f"Valor W (distancia): {w}")
print(f"Coordenadas de Pantalla (NDC): {p_ndc_persp}")
```
